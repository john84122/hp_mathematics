```python
import numpy as np
import pandas as pd

import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt

from tqdm import tqdm

from scipy.integrate import solve_ivp
```

```python

class lorenze_system():
    def __init__(self):
        self.phi = 10
        self.rho = 28
        self.beta = 8/3

    def change_rho(self, n_r):
        self.rho = n_r

    def dx_dt(self, x, y, z):

        return self.phi*(y-x)

    def dy_dt(self, x, y, z):
        return (x * (self.rho - z)) - y

    def dz_dt(self, x, y, z):
        return (x * y ) - (self.beta*z)

    def __call__(self, t, Y):
    
        x, y, z = Y

        return [self.dx_dt(x, y, z), self.dy_dt(x, y, z), self.dz_dt(x, y, z)]
```

```python
solutions_lst = []
pbar = tqdm(total = len(np.arange(5, 50, 0.25)), leave = False)
for rho in np.arange(5, 50, .5):

    lorenz = lorenze_system()

    lorenz.change_rho(rho)

    t_span = (0.0, 20.0)
    y0 = [0.0, 1.0, 0.0]
    solution = solve_ivp(lorenz, t_span, y0, dense_output=True, max_step= 0.01)
    solutions_lst.append((solution, rho))

    pbar.update(1)
```

```python
def create_pandata(sol_lst):

    dictionary = {
        "x_coord": [],
        "y_coord": [],
        "z_coord": [],
        "t_coord": [],
        "rho value": [],
    }

    for tup in sol_lst:
        dictionary["x_coord"] += list(tup[0].y[0])
        dictionary["y_coord"] += list(tup[0].y[1])
        dictionary["z_coord"] += list(tup[0].y[2])
        dictionary["t_coord"] += list(tup[0].t)
        dictionary["rho value"] += [tup[1]]*len(list(tup[0].y[0]))

    pandata = pd.DataFrame(dictionary)

    return pandata
```

```python
pandata = create_pandata(solutions_lst)
```
