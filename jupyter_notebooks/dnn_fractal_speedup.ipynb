```python
pandata_spec = pandata_speedup[pandata_speedup['machine'] == 'dgx0']
```

```python
pandata_spec.sort_values('time', inplace=True)
```

```python
sns.set_theme()

fig, ax = plt.subplots(figsize=(18, 5))

sns.barplot(ax = ax, x='Type', y='time', data=pandata_spec)

for container in ax.containers:
    ax.bar_label(container, padding=3)

ax.set_xlabel("Parallelization Framework", weight = "bold", size = 20)
ax.set_ylabel("Time (s)", weight = "bold", size = 20)
ax.set_title("DNN Fractal Generation Runtimes", weight = "bold", size = 25)
```
